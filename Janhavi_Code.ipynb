```python
#Question 2 (A)

Y = df["pass"].map({"yes":1,"no":0}) #binary encoding the categorical variables here where yes corresponds to 1 and no corresponds to 0. We have encoded the target variable.

feature_setA = df[["studytime", "absences"]]
feature_setB = df[["studytime", "absences", "Pstatus"]]

from sklearn.model_selection import train_test_split
train_index, test_index = train_test_split(df.index, test_size = 0.30,random_state = 42, stratify = Y) #taking the row indices of the data and randomly asisgning them into training and tetsing indices respectively. Used roww indices so that each Feature Set uses the same group of students to train and test because otherwise we wouldn't know whether one model performed better because of the additional predictor or because it got an easier group of students. Also stratify = Y ensures that the training and testing data have approximately the same proportion of Pass and Fail students.

y_train = Y.loc[train_index] #finding the target variables from the row indices which are in the train index
y_test = Y.loc[test_index] #did the same but for test index

feature_setA_train = feature_setA.loc[train_index] #finding the feature variables from the row indices which are in the train index
feature_setA_test = feature_setA.loc[test_index] #did the same but for test index

feature_setB_train = feature_setB.loc[train_index] #repeated the same process but for feature set B
feature_setB_test = feature_setB.loc[test_index]

#As seen above, there were quite a few missing values especially in important areas for e.g. in the studytimes feature. Hence, we'll handle these missing values by imputing them using the median value from the training data and not the mean (because it is sensitive to outliers).

studytime_median = feature_setA_train["studytime"].median() #found the median from the training dataset of feature set A and this median will be the same for feature set B as well

feature_setA_train_imputed = feature_setA_train.fillna({"studytime": studytime_median}) #imputed the missing values for each feature set and its corresponding training and test data; thereby handled the missing values
feature_setA_test_imputed = feature_setA_test.fillna({"studytime": studytime_median})
feature_setB_train_imputed = feature_setB_train.fillna({"studytime": studytime_median})
feature_setB_test_imputed = feature_setB_test.fillna({"studytime": studytime_median})

#now we have to encode "Pstatus" for feature set B
featuresetB_train_status = feature_setB_train["Pstatus"].map({"T":1,"A":0}) #binary encoded the categorical variable where T (living together) corresponds to 1 and A (apart) corresponds to 0
featuresetB_test_status = feature_setB_test["Pstatus"].map({"T":1,"A":0})

#standardizing the numerical predictors for logistic regression
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()

feature_setA_train_scaled = scaler.fit_transform(feature_setA_train_imputed[["studytime", "absences"]]) #calculated the mean and standard deviation using the training data and used those value to standardize the training data
feature_setA_test_scaled = scaler.transform(feature_setA_test_imputed[["studytime", "absences"]]) #used the same mean and standard deviation to standardize the test data

feature_setB_train_scaled = scaler.transform(feature_setB_train_imputed[["studytime", "absences"]]) #did the same for feature set B and used transform() this time because we can use the scaler from feature set A's training data since the same numerical columns are present in both feature sets.
feature_setB_test_scaled = scaler.transform(feature_setB_test_imputed[["studytime", "absences"]])

feature_setB_train_scaled1 = pd.DataFrame(feature_setB_train_scaled, columns =["studytime", "absences"],index = train_index) #StandardScaler() returns a NumpyArray so it was converted back into a dataframe making sure the column names are the same as before and the indixes of the training data remain
feature_setB_test_scaled1 = pd.DataFrame(feature_setB_test_scaled, columns =["studytime", "absences"],index = test_index) #applied the same process for the test data

feature_setB_train_scaled1["Pstatus"] = featuresetB_train_status #added the encoded "Pstatus" to each dataframe
feature_setB_test_scaled1["Pstatus"] = featuresetB_test_status

feature_setB_train_tree = pd.DataFrame({"studytime": feature_setB_train_imputed["studytime"], "absences": feature_setB_train["absences"], "Pstatus": featuresetB_test_status}, index = train_index) #for a decision tree, the numerical predictors don;t have to be standardized and hence we used the original numerical values in feature set B
feature_setB_test_tree = pd.DataFrame({"studytime": feature_setB_test_imputed["studytime"], "absences": feature_setB_test["absences"], "Pstatus": featuresetB_test_status}, index = test_index) #did the same for test
```

```python
#Question 2 (B)

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier

logistic_A = LogisticRegression(C=1, max_iter=2000) #fitted LogisticRegression according to the parameters listed in the question
logistic_A.fit(feature_setA_train_scaled, y_train) #basically teaching the model to predict Pass/Fail for a student based on the two standardized predictors

logistic_B = LogisticRegression(C=1, max_iter=2000) #did the same for feature set B
logistic_B.fit(feature_setB_train_scaled1, y_train) #here we used feature_setB_train_scaled1 i.e. the data with the "Pstatus" predictor so we can see whether adding this predictor will improve the model's performance or not

dt_A = DecisionTreeClassifier(max_depth=3, random_state=42) #applied the same concept to the decision tree
dt_A.fit(feature_setA_train_imputed[["studytime", "absences"]], y_train) #used orginal values and not standardized ones

dt_B = DecisionTreeClassifier(max_depth=3, random_state=42)
dt_B.fit(feature_setB_train_tree, y_train)
```

```python
#Question 2 (C)

from sklearn.metrics import accuracy_score
from sklearn.metrics import f1_score
from sklearn.metrics import confusion_matrix

pred_logistic_A = logistic_A.predict(feature_setA_test_scaled) #made predictions for each model
pred_logistic_B = logistic_B.predict(feature_setB_test_scaled1)
pred_dt_A = dt_A.predict(feature_setA_test_imputed[["studytime", "absences"]])
pred_dt_B = dt_B.predict(feature_setB_test_tree)

accuracy_logistic_A = accuracy_score(y_test, pred_logistic_A) #calculated the accuracy score for each model
accuracy_logistic_B = accuracy_score(y_test, pred_logistic_B)
accuracy_dt_A = accuracy_score(y_test, pred_dt_A)
accuracy_dt_B = accuracy_score(y_test, pred_dt_B)

f1_logistic_A = f1_score(y_test, pred_logistic_A, pos_label=1) #calculated the f1 score for each model. pos_label = 1 ensured that Pass was considered as the positive class
f1_logistic_B = f1_score(y_test, pred_logistic_B, pos_label=1)
f1_dt_A = f1_score(y_test, pred_dt_A, pos_label=1)
f1_dt_B = f1_score(y_test, pred_dt_B, pos_label=1)

results = pd.DataFrame({ #created a comparison table entailing all the accuracy and f1 scores
    "Model": ["Logistic Regression", "Logistic Regression", "Decision Tree", "Decision Tree"],
    "Features" : ["A", "B", "A", "B"],
    "Test Accuracy": [accuracy_logistic_A, accuracy_logistic_B, accuracy_dt_A, accuracy_dt_B],
    "Test F1": [f1_logistic_A, f1_logistic_B, f1_dt_A, f1_dt_B],
})

display(results.round(3)) #rounded the results to 3 decimal places

print(" ")
print(" ")

cm_logistic_B = confusion_matrix(y_test, pred_logistic_B, labels = [0,1]) #created confusion matrice for feature set B (Logistic Regression)
cm_logistic_B = pd.DataFrame(cm_logistic_B, columns = ["Predicted Fail", "Predicted Pass"], index = ["Actual Fail", "Actual Pass"])

cm_dt_B = confusion_matrix(y_test, pred_dt_B, labels = [0,1]) #did the same for feature set B (Decision Tree)
cm_dt_B = pd.DataFrame(cm_dt_B, columns = ["Predicted Fail", "Predicted Pass"], index = ["Actual Fail", "Actual Pass"])

combined_cm = pd.DataFrame({
    "Predicted Fail": [cm_logistic_B.loc["Actual Fail", "Predicted Fail"], cm_logistic_B.loc["Actual Pass", "Predicted Fail"], cm_dt_B.loc["Actual Fail", "Predicted Fail"], cm_dt_B.loc["Actual Pass", "Predicted Fail"]],
    "Predicted Pass": [cm_logistic_B.loc["Actual Fail", "Predicted Pass"], cm_logistic_B.loc["Actual Pass", "Predicted Pass"], cm_dt_B.loc["Actual Fail", "Predicted Pass"], cm_dt_B.loc["Actual Pass", "Predicted Pass"]]
}, index = ["Actual Fail — Logistic B", "Actual Pass — Logistic B", "Actual Fail — Tree B", "Actual Pass — Tree B"])

display(combined_cm) #printed the confusion matrices for each model
```

```python
#Question 3 (A)

incorrect_logistic = pd.Series(pred_logistic_B != y_test.values, index=y_test.index) #to identify one test observation that was incorrectly classified we need to compare the actual and predicted outcomes. Here, the predicted outcomes is a NumPy Array and so to compare the two, used .values to convert the actual outcomes into a NumPy array as well.
incorrect_dt = pd.Series(pred_dt_B != y_test.values, index=y_test.index) #applied the same process to identify one test observation that was incorrectly classified for decision tree

student_incorrect = incorrect_logistic | incorrect_dt #the condition is to find a student for which at least one model made an incorrect prediction and so even if one of the models (i.e. logistic regression or decision tree) makes an incorrect prediction, the test observation is considered incorrectly predicted.
student_incorrect_index = student_incorrect[student_incorrect].index[0] #from the above line we know student_incorrect returns only boolean values i.e. True or False and so we used that as an index to filter the Series to students who were incorrectly predicted.

student_incorrect_index #i.e. this is the index of the first student whose prediction was incorrect
```

```python
#Question 3 (B)

original_predictor = feature_setB_test.loc[student_incorrect_index] #retrived the original predictor value and actual outcome for the student at the 315th index (as found above)
actual_outcome = y_test.loc[student_incorrect_index]

pred_logisticB_series = pd.Series(pred_logistic_B, index = y_test.index) #to retreive predictions for each student, we need to convert the predictions in a Series which we have done here
pred_dtB_series = pd.Series(pred_dt_B, index = y_test.index)

logistic_prediction = pred_logisticB_series.loc[student_incorrect_index] #thereby was able to retreive the predictions through indexing and keeping the original dataframe indices the same.
dt_prediction = pred_dtB_series.loc[student_incorrect_index]

student_logistic = feature_setB_test_scaled1.loc[[student_incorrect_index]] #retrieving the student's data from each model
student_dt = feature_setB_test_tree.loc[[student_incorrect_index]]

prob_logistic = logistic_B.predict_proba(student_logistic) #finding the probabilities from each model
prob_dt = dt_B.predict_proba(student_dt)

pass_prob_logistic = prob_logistic[0,1] #retrieving the probability of passing from each model
pass_prob_dt = prob_dt[0,1]

print("Original Predictor Values:")
print(original_predictor)
print(" ")
print("Actual Outcome:", actual_outcome)
print(" ")
print("Logistic Regression Prediction:", logistic_prediction)
print(" ")
print("Decision Tree Prediction:", dt_prediction)
print(" ")
print("Logistic Regression Probability of Passing:", pass_prob_logistic)
print(" ")
print("Decision Tree Probability of Passing:", pass_prob_dt)
```

```python
#Question 3 (C)

c_logistic = pd.DataFrame(student_logistic)
c_dt = pd.DataFrame(student_dt)

c_logistic["Pstatus"] = 0 #earlier, the original Pstatus value was 1 which corresponds to T before encoding. now we want to check the same student's Pstatus had it been A so after encoding that would correspond to 0. Hence, here the changed logistic regression and changed decision tree is corresponding to 0.
c_dt["Pstatus"] = 0

new_prob_logistic = logistic_B.predict_proba(c_logistic)[0,1] #finding the pass probabilities from each model
new_prob_dt = dt_B.predict_proba(c_dt)[0,1]

change_logistic = new_prob_logistic - pass_prob_logistic #finding the change in pass probabilities
change_dt = new_prob_dt - pass_prob_dt

table = pd.DataFrame({ #presented the orginal and modified probabilities in a table as specified in the question
    "Model":["Logistic B", "Tree B"],
    "Original Label": ["T", "T"],
    "P(Pass) before": [pass_prob_logistic, pass_prob_dt],
    "P(Pass) after": [new_prob_logistic, new_prob_dt],
    "Change": [change_logistic, change_dt],
})

display(table.round(3)) #rounded the values in the table to 3 decimal places
```
